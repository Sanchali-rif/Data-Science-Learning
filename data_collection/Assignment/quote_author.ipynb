## From Quotes, extract all quotes & authors from all pages with life tag

```python
import requests
# fetch the data
URL="https://quotes.toscrape.com/"
res=requests.get(URL)
res.status_code
```

```python
# store the data
with open("scraped_data/data.html","w",encoding="utf-8") as f:
    f.write(res.text)
```

```python
from bs4 import BeautifulSoup
with open("scraped_data/data.html","r",encoding="utf-8") as f:
    content=f.read()
soup=BeautifulSoup(content,"lxml")
```

```python
all_info=[]
for i in soup.find_all("div", class_="quote"):
    tag_list=[]
    for j in i.select("span.text"):
        quote=j.get_text(strip=True) 
    for k in i.select("small.author"): 
        author=k.get_text(strip=True)  
    for l in (i.select("a.tag")):
        tag_list.append(l.get_text())
    all_info.append([quote,author,tag_list])
print(all_info)
```

## final answer for page-1

```python
for i in all_info:
    if "life" in i[2]:
        print(i[0],i[1])
```
