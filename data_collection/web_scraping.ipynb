## Fetching the data

```python
import requests
URL="http://scrapethissite.com/pages/simple/" # a webscraping site
res=requests.get(URL)
print(res.status_code) # returns HTTP status
```

```python
print(res.headers) # metadata like content type , server info etc
```

```python
if res.status_code==200: # always check status code
    print(res.text) # returns the html content of the web page
```

## Downloading & storing to file

```python
with open("scraped_data/data1.html","w", encoding="utf-8") as f:
    f.write(res.text) # write the file content in the file
```

## parsing the data - BeautifulSoup

```python
from bs4 import BeautifulSoup
with open ("scraped_data/data1.html","r", encoding="utf-8") as f:
    html_content=f.read()
    soup=BeautifulSoup(html_content,"lxml")
```

```python
soup.find("h1") # first matching element -- 1st h1 tag
```

```python
all_h3=soup.find_all("h3") # all information about h3 tags
for i in all_h3:
    name=i.get_text(strip=True) # all the countries are extracted from the header
    #print(i.find_parent("div")["class"]) # find h3's div parent's class
    #get all the population value--
    print(i.find_next("div").select("span.country-population")[0].get_text(strip=True))
```

```python
# attribute selector
all_h3=soup.find_all("h3") # all information about h3 tags
for i in all_h3:
    print(i.attrs) # all attributes of h3
```
