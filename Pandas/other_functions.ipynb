```python
import pandas as pd
df=pd.read_csv("raw_data (1).csv")
df
```

```python
# Grouping & aggregation of data
df.groupby("country")["income"].mean() # mean income for each country
```

```python
df.groupby("country")["income"].min() # minimum income of each country
```

```python
df.groupby("country")["income"].max() # maximum income of each country
```

```python
df.groupby("gender")["income"].mean() # mean income of each gender
```

```python
df.groupby("country")["income"].agg(["mean","min","max"]) #multiple function together
# or
df.groupby("country")["income"].agg(avg_salary="mean",min_salary="min",max_salary="max")
```
