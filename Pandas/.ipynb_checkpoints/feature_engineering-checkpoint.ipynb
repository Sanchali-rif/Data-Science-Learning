Feature engineering

```python
#data transformation
import pandas as pd
df=pd.read_csv("raw_data (1).csv")
df
```

```python
# Applies a function to each row, column, or element.
df2=df.copy()
df2["tax"]=df2["income"].apply(lambda x:"20%" if x>=50000 else "10%")
df2
```

```python
# Maps each value in a Series using a dictionary or function.
gender_map={"Male":"M","Female":"F","Unknown":"U"}
df2["gender"]=df2["gender"].map(gender_map)
df2
```

```python
# Creates new columns (or modifies existing ones) and returns a new DataFrame.
df2.assign(new_income=df2["income"]*1.1) 
```

```python
# Replaces specific values with new ones.
df2["country"]=df2["country"].replace("USA","US")
```

```python
df2
```

```python
df2.columns=["Id","Name","Age","Country","Gender","Income","Tax"] #rename all coulmns
df2
```

```python
df2.rename(columns={"Income":"salary"}) # rename perticular columns
```

```python
df2.rename(index={1:"First"}) #rename row name
```

```python
df2["Income"].sort_values() # sorting based on income
df2.sort_values("Income")
```

```python
df2.sort_values("Income",ascending=False) # descending sorting
```

```python
df2=df2.fillna(50)
sorted_df=df2.sort_values(["Income","Age"]) #s sort based on income , if two income are same then sortbased on age
sorted_df
```

```python
sorted_df.sort_index() # sort index
```

```python
sorted_df.reset_index() # reset the index with preserved values
```

```python
sorted_df.reset_index(drop=True) # removes the preserved value
```

```python
sorted_df["Ranking"]=sorted_df["Income"].rank() # assign rank based on Income
sorted_df                                       # 1 => lowest rank
```

```python
sorted_df["Ranking"]=sorted_df["Income"].rank(ascending=False)  # 1 => heighest rank
sorted_df  
```

```python
sorted_df["Ranking"]=sorted_df["Income"].rank(ascending=False,method="dense")  # assign rounded figuer instead of .5 in the same values
sorted_df  
```

```python
sorted_df["Ranking"]=sorted_df["Income"].rank(ascending=False,method="min")  # assing lowest rank in the same values
sorted_df  
```

```python
sorted_df["Ranking"]=sorted_df["Income"].rank(ascending=False,method="max")  # assign heighest rang in the same values
sorted_df  
```
