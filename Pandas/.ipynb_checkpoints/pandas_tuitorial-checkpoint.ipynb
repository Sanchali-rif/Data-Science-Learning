```python
import pandas as pd
```

```python
#datafram
info={
    "Name" : ["sanchali","rima","debu","adrija"],
    "CGPA" : [9.5,7.5,8.9,9.7]
}
df=pd.DataFrame(info) #converted the info dictionary into 2d labelled array / table
print(df)
```

```python
df
```

```python
#series
s=pd.Series(["rima","puma","nina","uma","lila"])
print(s)
#index based acess
print(s[2])
```

```python
#custom lebels
marks=pd.Series([20,19,21,20,22],index=["sanchali","nina","dev","gopal","akash"])
print(marks)

#lebel based access
print(marks["dev"])
print(marks.index)
```

```python
# Series properties: vectorised operations are allowed
s1=pd.Series([1,2,3,4,5])
s2=pd.Series([10,20,30,40,50])
print(s1+s2)
#mutable values
s1[0]=10
print(s1+s2)
# immutable size - cannot add or delete any element but can be stored in new series
changed_s1=s1.drop(0)
print(s1)
print(changed_s1)
```

```python
#datafram-can be created by disctionary of lists
info={
    "name":["Sanchali","Adam","Eve","Bob"],
    "age":[20,21,22,19],
    "GPA":[9.5,8.6,7.2,10.0]
}

students=pd.DataFrame(info)
students
```

```python
print(students.index) #row operatison
print(students.columns) #column operation
```

```python
#can also be created by list of lists
info=pd.DataFrame([["sanchali",20],["Uma",21],["shila",40]],columns=["name","age"])
info
```

```python
#can be created by NumPy arrays

import numpy as np
np_arr=np.array([[1,2,3],[4,5,6]])
df=pd.DataFrame(np_arr,columns=["A","B","C"])
df
```
