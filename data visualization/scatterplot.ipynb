```python
import matplotlib.pyplot as plt
```

```python
# Scatter Plots

people = ["Person A", "Person B", "Person C", "Person D", "Person E",
          "Person F", "Person G", "Person H", "Person I", "Person J"]
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
blood_pressure = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]
# correlation of age and blood pressure-
plt.scatter(age,blood_pressure)
plt.title("Age Vs Bp")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.grid() # optional - for grid view
```

```python
# red if bp >135 and green if bp<135
people = ["Person A", "Person B", "Person C", "Person D", "Person E",
          "Person F", "Person G", "Person H", "Person I", "Person J"]
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
BP = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]
colours=["green" if x<135 else "red" for x in BP]
plt.scatter(age,BP,color=colours)
plt.title("Age Vs Blood pressure")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
```

```python
# Transparency of the point
people = ["Person A", "Person B", "Person C", "Person D", "Person E",
          "Person F", "Person G", "Person H", "Person I", "Person J"]
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
BP = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]
plt.scatter(age,blood_pressure,alpha=0.5) # 0-1
plt.title("Age Vs Bp")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
```

```python
# colormap
people = ["Person A", "Person B", "Person C", "Person D", "Person E",
          "Person F", "Person G", "Person H", "Person I", "Person J"]
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
BP = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]
plt.scatter(age,BP,cmap="OrRd",c=BP) # c defines the range of the color
plt.title("Age Vs Bp")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.colorbar(label="BP") # shows the color bar
```

```python
#annotations
people = ["Person A", "Person B", "Person C", "Person D", "Person E",
          "Person F", "Person G", "Person H", "Person I", "Person J"]
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
blood_pressure = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]

plt.scatter(age,blood_pressure)
for i in range (len(people)):
    # added the people's name with their BP point
    plt.annotate(people[i],xy=(age[i],blood_pressure[i]),xytext=(age[i]-2,blood_pressure[i]+1))
plt.title("Age Vs Bp")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.xlim(18,max(age)+5) # expanded the graph in x axis
plt.ylim(min(blood_pressure),max(blood_pressure)+5) # expanded the graph in y axis
```
