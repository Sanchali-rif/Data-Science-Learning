```python
import matplotlib.pyplot as plt
```

```python
#bar chart 
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]

years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]
plt.bar(years,oscar_revenue,color="green") # to create bar graph
plt.title("Oscar movies Revenue in each year")
plt.xlabel("year")
plt.ylabel("Revenue in ($M)")
# add extra details
for i in range(len(oscar_revenue)):
    #adding revenue of each movie at the top of the graph
    plt.text(years[i],oscar_revenue[i]+10,str(oscar_revenue[i]),ha="center") 
    # oscar_revenue[i]+10 => +10 is to create a gap between the graph and the text

# to increase graph limits
# plt.ylim(0,max(oscar_revenue)+100) -- for y axix limit
# plt.xlim(min(years),max(years)+100) -- for X axix limit
```

## Compare 2 sets of Data

```python
# Print revenue of these movies per year
import numpy as np
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
oscar_revenue = [1005, 170, 427, 133, 232]

non_oscar_movies = ["Slumdog Millionaire", 
                    "Avatar", "Inception", 
                    "Hugo", 
                    "Lincoln"
]
non_oscar_revenue = [378, 2788, 829, 185, 275] 

years = [2008, 2009, 2010, 2011, 2012]

plt.title("Oscar Movies Vs Non Oscare Movies Revenue")
plt.xlabel("year")
plt.ylabel("Revenue ($M)")

x=np.arange(len(years)) # array([0, 1, 2, 3, 4])
width=0.4

plt.bar(x-width/2,oscar_revenue,width,label="Oscar Movies")
plt.bar(x+width/2,non_oscar_revenue,width,label="Non-Oscar Movies")
plt.legend()
plt.xticks(x,years) # replace the values of x in x axis with years
```

```python
# horizontal bar charts
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
oscar_revenue = [1005, 170, 427, 133, 232]
plt.barh(oscar_movies,oscar_revenue)
plt.xlabel("Revenue ($M)")
plt.ylabel("Oscar Movies")
plt.title("Oscar Movies Vs Revenue")
```
