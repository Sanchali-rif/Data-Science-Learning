```python
import matplotlib.pyplot as plt
data=[7,8,5,6,9,4,10,12,15]
plt.boxplot(data) #
plt.grid(True)
```

```python
# random multiple dataset
import numpy as np
group1=np.random.normal(50,10,100)
group2=np.random.normal(60,15,100)
plt.boxplot([group1,group2],tick_labels=["group1","group2"])
plt.grid()
plt.show()
```

```python
#horizontal graph
import matplotlib.pyplot as plt
data=[7,8,5,6,9,4,10,12,15]
plt.boxplot(data,vert=False) #
plt.grid(True)
```

```python
# show mean
import matplotlib.pyplot as plt
data=[7,8,5,6,9,4,10,12,15]
plt.boxplot(data,showmeans=True) #
plt.grid(True)
```

```python
# wisker length
import matplotlib.pyplot as plt
data=[7,8,5,6,9,4,10,12,15]
plt.boxplot(data,whis=2.0) # changed the wisker length
plt.grid(True)
```
