```python
import seaborn as sns
import matplotlib.pyplot as plt
tips=sns.load_dataset("tips")
tips.head()
```

```python
# for ploting graph --> use seaborn
fig,ax=plt.subplots()
sns.lineplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    ax=ax # connecting matplotlib and seaborn for customization for perticular plot no the whole canvas(fig)
)
```

```python
# customization --> use matplotlib
fig,ax=plt.subplots()
sns.lineplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    marker="o",
    errorbar=None, # removes confidence interval
    ax=ax # connecting matplotlib and seaborn for customization
)
ax.set_title("avg bill value for each day")
ax.set_xlabel("Days")
ax.set_ylabel("Total Bill Amount")
fig.tight_layout
```
