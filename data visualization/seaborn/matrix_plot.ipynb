```python
# heatmap
import seaborn as sns
import matplotlib.pyplot as plt
flights=sns.load_dataset("flights")
```

```python
flights
```

```python
flights.pivot(index="month",columns="year",values="passengers")
```

```python
flight_pivot=flights.pivot(index="month",columns="year",values="passengers")
sns.heatmap(
    flight_pivot,
    cmap="coolwarm"
)
```

```python
flight_pivot=flights.pivot(index="month",columns="year",values="passengers")
sns.heatmap(
    flight_pivot,
    cmap="coolwarm", # for color scheme
    annot=True, # annotate the number of each plot
    fmt="d" # formate string
)
plt.title("passangers heatmap")
plt.tight_layout()
```
