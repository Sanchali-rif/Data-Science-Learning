```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# 1st set theme 
sns.set_theme()
```

```python
sns.get_dataset_names()
```

```python
tips=sns.load_dataset("tips")
print(type(tips)) # it is a dataframe,so we can perform pandas operations here
```

```python
tips.head()
```

```python
sns.set_theme(style="whitegrid")
sns.relplot( # default relational plot is scatter plot
    data=tips, # pass the dataset
    x="total_bill", #directly pass the columns required 
    y="tip"
)
```

```python
# can also heighlight smoker and non smoker datapoints
sns.relplot( 
    data=tips, 
    x="total_bill", 
    y="tip",
    hue="smoker", # it is done by hue
    size="size", # can also differentiate via size
    style="smoker" # give style based on who is smoker and who is non smoker
)
```

```python
# line plot
sns.set_theme(style="darkgrid")
x_axis=[1,2,3,4,5,6]
y_axis=[i**3 for i in x_axis]
sns.relplot(
    x=x_axis,
    y=y_axis,
    kind="line" # change the kind to line
)
```
