```python
import matplotlib.pyplot as plt
```

```python
expenses=["Salaries","Rent","Marketing","R&D","Miscellaneous"]
amounts=[500,150,120,100,50]
plt.pie(amounts,labels=expenses)
```

```python
# can use styles also
expenses=["Salaries","Rent","Marketing","R&D","Miscellaneous"]
amounts=[500,150,120,100,50]
plt.style.use("dark_background")
plt.pie(amounts,labels=expenses)
plt.style.available
```

## customization

```python
# shows the percentage of each slice
expenses=["Salaries","Rent","Marketing","R&D","Miscellaneous"]
amounts=[500,150,120,100,50]
plt.pie(amounts,labels=expenses,autopct="%1.1f%%")
plt.style.use("default")
plt.show()
```

```python
# change the color of each slice
expenses=["Salaries","Rent","Marketing","R&D","Miscellaneous"]
amounts=[500,150,120,100,50]
colours=["blue","yellow","orange","green","pink"]
plt.pie(amounts,labels=expenses,autopct="%1.1f%%",colors=colours)
```

```python
# wedgeprops -- for multiple style changes
expenses=["Salaries","Rent","Marketing","R&D","Miscellaneous"]
amounts=[500,150,120,100,50]
plt.pie(amounts,labels=expenses,autopct="%1.1f%%",
        wedgeprops={
        "edgecolor":"black", # black border
         "linewidth":1, # width of line
         "linestyle":"--", # style of line
         "alpha":1, # fully visible , not transparent
         "fill":False # no filling of colour in the slice
        } 
       )
```

```python
# highlight one slice -- slice will be pulled out
expenses=["Salaries","Rent","Marketing","R&D","Miscellaneous"]
amounts=[500,150,120,100,50]
explode=[0,0.4,0,0.2,0] # R&D and Rent is pulled out
plt.pie(amounts,labels=expenses,autopct="%1.1f%%",
        wedgeprops={
        "edgecolor":"black", 
         "linewidth":1, 
         "linestyle":"--", 
        },
        explode=explode
       )
```

```python
# rotate the pei chart 
expenses=["Salaries","Rent","Marketing","R&D","Miscellaneous"]
amounts=[500,150,120,100,50]
explode=[0,0.4,0,0.2,0] # R&D and Rent is pulled out
plt.pie(amounts,labels=expenses,autopct="%1.1f%%",
        wedgeprops={
        "edgecolor":"black", # black border
         "linewidth":1, # width of line
         "linestyle":"--", # style of line
        },
        explode=explode,
        startangle=45 # 45 degree of rotation
       )
```

```python
# for 3D effect 
expenses=["Salaries","Rent","Marketing","R&D","Miscellaneous"]
amounts=[500,150,120,100,50]
explode=[0,0,0,0.2,0] # R&D and Rent is pulled out
plt.pie(amounts,labels=expenses,autopct="%1.1f%%",
        wedgeprops={
        "edgecolor":"black", 
         "linewidth":1, 
         "linestyle":"--", 
        },
        explode=explode,
        shadow=True # added a shadow
       )
```
