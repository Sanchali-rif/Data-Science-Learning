# For the following dataset generated using NumPy:

```python
np.random.normal(loc=70, scale=10, size=200)
```
- Plot a histogram of scores.
- Draw a vertical line for the mean & try to interpret the results.

````python
import numpy as np
import matplotlib.pyplot as plt
num=np.random.normal(loc=70, scale=10, size=200)
plt.hist(num)
plt.title("histogram of random normal distribution")
plt.xlabel("Scores")
plt.ylabel("frequency")
m=np.mean(num)
plt.axvline(
    x=m,
    color="red",
    linestyle="--",
    label="mean",
    linewidth=2
)
````
