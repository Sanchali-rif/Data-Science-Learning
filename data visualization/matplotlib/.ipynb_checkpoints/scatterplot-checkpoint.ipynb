```python
import matplotlib.pyplot as plt
```

```python
# Scatter Plots

people = ["Person A", "Person B", "Person C", "Person D", "Person E",
          "Person F", "Person G", "Person H", "Person I", "Person J"]
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
blood_pressure = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]
# correlation of age and blood pressure-
plt.scatter(age,blood_pressure)
plt.title("Age Vs Bp")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.grid() # optional - for grid view
```

## customization

```python
# colour of the dots
people = ["Person A", "Person B", "Person C", "Person D", "Person E",
          "Person F", "Person G", "Person H", "Person I", "Person J"]
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
blood_pressure = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]

plt.scatter(age,blood_pressure,color="orange") # colour changed to orange
plt.title("Age Vs Bp")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
```

```python
# red if bp >135 and green if bp<135
people = ["Person A", "Person B", "Person C", "Person D", "Person E",
          "Person F", "Person G", "Person H", "Person I", "Person J"]
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
BP = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]
colours=["green" if x<135 else "red" for x in BP]
plt.scatter(age,BP,color=colours)
plt.title("Age Vs Blood pressure")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
```

```python
# size of the point
people = ["Person A", "Person B", "Person C", "Person D", "Person E",
          "Person F", "Person G", "Person H", "Person I", "Person J"]
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
BP = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]
plt.scatter(age,blood_pressure,s=BP) # as the Bp increases the point gets bigger
plt.title("Age Vs Bp")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
```

```python
# shape of the point
people = ["Person A", "Person B", "Person C", "Person D", "Person E",
          "Person F", "Person G", "Person H", "Person I", "Person J"]
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
BP = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]
plt.scatter(age,blood_pressure,color="pink",marker=">") # as the Bp increases the point gets bigger
plt.title("Age Vs Bp")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
```

```python
# Transparency of the point
people = ["Person A", "Person B", "Person C", "Person D", "Person E",
          "Person F", "Person G", "Person H", "Person I", "Person J"]
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
BP = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]
plt.scatter(age,blood_pressure,alpha=0.6) # 0-1
plt.title("Age Vs Bp")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
```
