```python
import matplotlib.pyplot as plt
import numpy as np
np.random.seed(0) # starting point- same sequence
score=np.random.normal(70, 10, 100) # (mean, standard_deviation, number_of_values) also normal distribution
plt.hist(score)
plt.xlabel("marks")
plt.ylabel("frequency of the student")
```

## costumization

```python
# colour and edge colour
np.random.seed(0) 
score=np.random.normal(70, 10, 100) 
plt.hist(score,color="yellow",edgecolor="orange")
plt.xlabel("marks")
plt.ylabel("frequency of the student")
```

```python
# can change the range/bin 
np.random.seed(0) 
score=np.random.normal(70, 10, 100) 
plt.hist(score,edgecolor="black",bins=5) # this is default range
plt.xlabel("marks")
plt.ylabel("frequency of the student")
```

```python
# we can also customize range also
marks = [
    25, 38, 42, 55, 61, 67, 72, 78, 83, 88,
    91, 95, 47, 53, 59, 64, 69, 74, 79, 85,
    92, 97, 33, 45, 58, 62, 71, 76, 81, 89
]
bin_value=[0,30,70,90,100] # continous range -- 0-30,30-70...
plt.hist(marks,edgecolor="black",bins=bin_value)
plt.xlabel("marks")
plt.ylabel("frequency of the student")
plt.title("distribution of marks of students")
```

```python
# multiple datasets on histogram
legit_transactions = [
    2.99, 5.49, 8.99, 12.50, 14.99, 19.99, 23.45, 29.99, 34.99, 39.50,
    45.00, 49.99, 55.25, 60.00, 75.99, 89.99, 120.50, 150.00, 199.99,
    249.99, 300.75, 450.00, 600.00, 850.00, 1280.00
]

fraud_transactions = [
    50, 100, 150, 200, 300, 500, 500, 750, 1000, 1000,
    1200, 1500, 1800, 2000, 2500, 3000, 3000
]
plt.hist(fraud_transactions,color="red",edgecolor="black",bins=10,alpha=0.6,label="fraud transactions")
plt.hist(legit_transactions,color="green",edgecolor="black",bins=10,alpha=0.6,label="legit transactions")
plt.xlabel("Amount in dollar")
plt.ylabel("Frequency")
plt.title("transaction amount distribution")
plt.grid(True)
plt.legend()
```

```python
# drawing a verticle line to show the cutoff of the marks
marks = [
    25, 38, 42, 55, 61, 67, 72, 78, 83, 88,
    91, 95, 47, 53, 59, 64, 69, 74, 79, 85,
    92, 97, 33, 45, 58, 62, 71, 76, 81, 89
]
bin_value=[0,30,70,90,100] # continous range -- 0-30,30-70...
plt.hist(marks,edgecolor="black",color="skyblue",bins=bin_value)
plt.axvline(
    x=33,
    color="red",
    linestyle="--",
    label="passing marks(33)",
    linewidth=2
)
plt.xlabel("marks")
plt.ylabel("frequency of the student")
plt.title("distribution of marks of students")
```
