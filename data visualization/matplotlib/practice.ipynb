```python
# Weekly temperature
import matplotlib.pyplot as plt
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]

cities = ["New York", "London", "Delhi", "Tokyo"]

temperatures = [
    [22, 23, 21, 24, 25],  # New York
    [18, 19, 17, 20, 21],  # London
    [30, 32, 31, 33, 34],  # Delhi
    [25, 26, 24, 27, 28]   # Tokyo
]
fig,ax=plt.subplots(2,2)
city_num=0
for i in range(2): # row
    for j in range(2): # column
        ax[i][j].plot(days,temperatures[city_num],marker="o")
        ax[i][j].set_title(cities[city_num])
        city_num=city_num+1
fig.suptitle("weekly weather report")
fig.supylabel("Temperature") # xlabel
fig.supxlabel("Days") # ylabel
fig.tight_layout()
```

```python
# simpler version
fig,axes=plt.subplots(2,2) # axes is 2
axes=axes.flatten() # 2D->1D 
for i,ax in enumerate(axes): # enumerate func at 1D array returns - idx and val at idx
    ax.plot(days,temperatures[i],marker="o")
    ax.set_title(cities[i])

fig.tight_layout()
```
