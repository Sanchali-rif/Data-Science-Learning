```python
import matplotlib.pyplot as plt
```

```python
#bar chart 
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]

years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]
plt.bar(years,oscar_revenue,color="green") # to create bar graph
plt.title("Oscar movies Revenue in each year")
plt.xlabel("year")
plt.ylabel("Revenue in ($M)")
# add extra details
for i in range(len(oscar_revenue)):
    #adding revenue of each movie at the top of the graph
    plt.text(years[i],oscar_revenue[i]+10,str(oscar_revenue[i]),ha="center") 
    # oscar_revenue[i]+10 => +10 is to create a gap between the graph and the text

# to increase graph limits
# plt.ylim(0,max(oscar_revenue)+100) -- for y axix limit
# plt.xlim(min(years),max(years)+100) -- for X axix limit
```
