```python
# function to load the data
def load(filename):
    with open(filename,"r") as f:
        stored_data=json.load(f)
    return stored_data
```

```python
data=load("store_data.json")
print(data)
```

```python
# clean our data
def clean_data(filename):
    clean_data=[]
    unique_users=set()
    text_to_num={"zero":0,"one":1,"two":2,"three":3,"four":4,"five":5}
    for user in filename:
        #clean rating - data consistency
        raw_rating=str(user["rating"]).strip().lower() #removes space from starting and ending , convert into lowercase string
        if(raw_rating in text_to_num):
            raw_rating=text_to_num[raw_rating]
        user["rating"]=raw_rating

        #handle missing value=>age
        raw_age=user.get("age")
        if(raw_age==None):
            user["age"]=None

        #deduplication
        if(user["name"].strip() in unique_users):
            continue
        unique_users.add(user["name"])
        clean_data.append(user)
    return clean_data
```

```python
data=clean_data(data)
print(data)
```

```python
# get meaningful insights
def get_insights(data):
    # for avg rating
    total_ratings=0
    for user in data:
        total_ratings=total_ratings+float(user["rating"]) 
    print(f"the avg rating of the data is {total_ratings/len(data)}")

    #percentage of users with poor rating (less than 3)
    total_ratings=0
    less_rating=0
    for user in data:
        total_ratings=total_ratings+float(user["rating"]) 
        if (float(user["rating"])<3):
            less_rating=less_rating+1
    percentage=(less_rating/len(data))*100
    print(f"the percentage of poor rating of the data is {percentage}%")
```

```python
get_insights(data)
```

```python
# Build a product recommendation feature
def get_recommendations(data):
    recommendation_list=[]
    for user in data:
        recommendation={}
        recommendation["name"]=user["name"]
        if (float(user["rating"])>=4):
            recommendation["Brand"]="Apple"
        else:
            recommendation["Brand"]="samsung"
        recommendation_list.append(recommendation)
    print(recommendation_list)
```

```python
get_recommendations(data)
```
