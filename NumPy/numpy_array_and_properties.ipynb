```python
import numpy as np
```

```python
arr=np.array([1,2,3,4])
print(arr , type(arr) , arr.dtype) #all will be numners

arr2=np.array([1,2,3,4,"sanchali"])
print(arr2 , type(arr2) ,arr2.dtype) # all are strings because numpy arrays are homogenous
```
