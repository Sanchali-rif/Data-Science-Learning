```python
import numpy as np
import time
```

```python
arr=np.array([1,2,3,4,5]) #converting a list into ndarray
print(arr)
print(type(arr))
```

```python
#execution time of python list Vs numpy array
size=10_000_000
py_list= list(range(size)) #python list
start=time.time()
sq_list=[i*i for i in py_list]
end=time.time()
print(f"python list time = {end-start}")

np_arr=np.array(py_list) #numpy array
start=time.time()
sq_arr=np_arr**2  #vectorization- the operation get performed in the entire array at the same time
end=time.time()
print(f"numpy array time = {end-start}")
```

```python
#compare memory
import sys
print(f"python list size = {sys.getsizeof(py_list)* len(py_list)} bytes")

print(f"numpy array size = {np_arr.nbytes} bytes")
```
